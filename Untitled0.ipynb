```python
 # Water Quality & Safety Analyzer - Phase 1
# Global Standards: WHO Safe Water Thresholds

def analyze_water_sample(sample_id, location, ph, turbidity, tds):
    print(f"\n--- EVALUATING: {sample_id} ({location}) ---")
    warnings = []

    # 1. pH Level Check (Safe Range: 6.5 - 8.5)
    if ph < 6.5:
        warnings.append(f"pH level is too low ({ph}) -> Acidic Water")
    elif ph > 8.5:
        warnings.append(f"pH level is too high ({ph}) -> Alkaline Water")
    else:
        print(f"  [PASSED] pH Level: {ph}")

    # 2. Turbidity Check (Safe: < 5.0 NTU)
    if turbidity > 5.0:
        warnings.append(f"Turbidity is high ({turbidity} NTU) -> Contaminated/Cloudy")
    else:
        print(f"  [PASSED] Turbidity: {turbidity} NTU")

    # 3. Total Dissolved Solids (TDS) Check (Safe: < 500 ppm)
    if tds > 500:
        warnings.append(f"TDS is high ({tds} ppm) -> High Chemical/Mineral Content")
    else:
        print(f"  [PASSED] TDS: {tds} ppm")

    # Final Safety Status
    if warnings:
        print("  STATUS: [UNSAFE / WARNING]")
        for w in warnings:
            print(f"   - ALERT: {w}")
        return False
    else:
        print("  STATUS: [SAFE FOR DRINKING]")
        return True

# Sample Dataset
samples = [
    {"id": "W_001", "location": "North District Well", "ph": 7.2, "turbidity": 3.1, "tds": 250},
    {"id": "W_002", "location": "Industrial Zone River", "ph": 5.1, "turbidity": 8.4, "tds": 620},
    {"id": "W_003", "location": "Central Reservoir", "ph": 7.8, "turbidity": 1.2, "tds": 180}
]

# Run Analysis
safe_count = 0
for sample in samples:
    is_safe = analyze_water_sample(
        sample["id"], sample["location"], sample["ph"], sample["turbidity"], sample["tds"]
    )
    if is_safe:
        safe_count += 1

print("\n" + "="*45)
print(f"SUMMARY REPORT: {safe_count}/{len(samples)} samples are safe for drinking.")
print("="*45)

```
